## CatBoost

```python
!pip install catboost
import pandas as pd
from catboost import CatBoostClassifier

# Base de Dados Fictícia
dados = pd.DataFrame({
    'Idade': [25, 45, 30, 50, 23, 40],
    'Profissao': ['TI', 'Medico', 'TI', 'Advogado', 'Estudante', 'Vendas'], # Texto Bruto
    'Comprou': [0, 1, 0, 1, 0, 1]
})

X = dados.drop('Comprou', axis=1)
y = dados['Comprou']

# 1. Avisamos ao modelo quais colunas são texto (SEM LABEL ENCODER!)
colunas_texto = ['Profissao']

# 2. Treinamento Nativo
modelo_cat = CatBoostClassifier(iterations=30, depth=3, learning_rate=0.1, verbose=0)
modelo_cat.fit(X, y, cat_features=colunas_texto)

print("✅ CatBoost treinado com sucesso processando texto nativamente!")
```

## LightGBM

```python
!pip install lightgbm
import pandas as pd
from lightgbm import LGBMClassifier

dados = pd.DataFrame({
    'Idade': [25, 45, 30, 50, 23, 40],
    'Profissao': ['TI', 'Medico', 'TI', 'Advogado', 'Estudante', 'Vendas'],
    'Comprou': [0, 1, 0, 1, 0, 1]
})

X = dados.drop('Comprou', axis=1)
y = dados['Comprou']

# 1. O truque do LightGBM: Converter a coluna de texto para o tipo 'category'
X['Profissao'] = X['Profissao'].astype('category')

# 2. Treinamento
# min_child_samples=2 é usado APENAS para bases pequenas/didáticas não travarem
modelo_lgbm = LGBMClassifier(n_estimators=30, min_child_samples=2, verbose=-1, random_state=42)
modelo_lgbm.fit(X, y)

print("✅ LightGBM treinado com sucesso utilizando variáveis category do Pandas!")
```
