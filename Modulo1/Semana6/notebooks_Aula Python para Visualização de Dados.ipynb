## Primeiro exemplo
No exemplo abaixo, usamos dois pacotes de funcionalidades python (Numpy e Matplotlib) para criar alguns dados aleatórios e visualizá-los.

```python
from matplotlib import pyplot as plt
import numpy as np

# Generate 100 random data points along 3 dimensions
x, y, scale = np.random.randn(3, 100)
fig, ax = plt.subplots()

# Map each onto a scatterplot we'll create with Matplotlib
ax.scatter(x=x, y=y, c=scale, s=np.abs(scale)*500)
ax.set(title="Some random data, created with JupyterLab!")
plt.show()
```

## Estruturas de decisão em Python
### Use 'if', 'elif' e 'else' para executar comandos de acordo com condições.

```python
# Estruturas de decisão em Python
numero = int(input("Digite um número"))
if numero > 0:
    print(f"O número {numero} é positivo")
    print("Até mais")
elif numero < 0:
    print(f"O número {numero} é negativo")
    print("Até menos")
else:
    print(f"O número {numero} é zero")
    print("Até nada")
    
print("Fim")
```

## Estruturas de repetição
### Itere sobre sequências ou coleções, como ranges ou listas listas por exemplo, usando o comando 'for'.
### Repita enquanto uma condição for satisteita (True), usando um comando 'while'.

```python
for i in range(0, 10):
    print(i)
    
for i in range(2, 10, 3):
    print(i)

cont = 0
while cont < 10:
    print("Cont:", cont)
    cont +=1
```

## Funções
### Defina blocos reutilizáveis ​​com entradas/saídas claras para manter o código modular e testável.

```python
def square_evens(numbers):
    evens = [x ** 2 for x in numbers if x %2 == 0]
    return evens
```

```python
result = square_evens([1, 2, 3, 4, 5, 6])
print(result)
```

## Listas
### Crie listas (e outras coleções) de forma concisa, filtrando e transformando dados em uma única expressão legível.

```python
# Lista vazia
minha_lista_vazia = []

# Lista com elementos
minha_lista = [1, 2, 3, "a", "b", True]

# Lista de listas
minha_lista_de_listas = [[1, 2, 3],[3], [4, 5, 6]]

display(minha_lista_vazia)

display(minha_lista)

display(minha_lista_de_listas)

# Acessando o primeiro elemento (índice 0)
primeiro_elemento = minha_lista[0]  # Resultado: 10

print(primeiro_elemento)

# Acessando o último elemento (índice -1)
ultimo_elemento = minha_lista[-1]  # Resultado: 50

print(ultimo_elemento)

# Acessando um intervalo de elementos (slicing)
sub_lista = minha_lista[1:4]  # Resultado:
print(sub_lista)

# Alterando o valor do segundo elemento (índice 1)
#minha_lista[1] = 25
#print(minha_lista)

# Adicionando um elemento ao final da lista
minha_lista.append(60)
print(minha_lista)

# Removendo um elemento pelo valor
minha_lista.remove(3)
print(minha_lista)

# Removendo um elemento pelo índice
del minha_lista[0]
print(minha_lista)

# --- Listas: Guardando vários valores em uma caixa ---
alturas = [1.70, 1.85, 1.62, 1.75, 1.90]

# --- Funções: Comandos que executam ações ---
# A estrutura é sempre: nome_da_funcao(dado_de_entrada)
len(alturas)             # Conta quantos itens há na lista
sum(alturas)             # Soma todos os valores da lista
```

## Dicionários
### Diferente das listas (que usam índices numéricos), os dicionários armazenam dados em pares de chave: valor. Eles são ideais para representar dados estruturados, como informações de um país ou configurações de um gráfico.

```python
# Dicionários
meu_dicionario = {
    "nome": "João",
    "idade": 30,
    "cidade": "São Paulo",
    "hobbies": ["leitura", "esportes"]
}

print(meu_dicionario["nome"])  # Saída: João
print(meu_dicionario["idade"]) # Saída: 30

#meu_dicionario["idade"] = 31  # Modificando o valor da chave "idade"
#meu_dicionario["profissao"] = "Engenheiro" # Adicionando um novo par chave-valor

print(meu_dicionario)
```

```python
# 1. Criando um dicionário (chave: valor)
dados_pais = {
    "nome": "Brasil",
    "expectativa_vida": 75.5,
    "continente": "América"
}

# 2. Acessando um valor através da chave
print(dados_pais["nome"])  # Saída: Brasil

# 3. Adicionando ou atualizando um par chave-valor
dados_pais["populacao_milhoes"] = 214.3
dados_pais["expectativa_vida"] = 76.0  # Atualiza o valor existente

# 4. Verificando se uma chave existe
if "continente" in dados_pais:
    print("A chave 'continente' está presente.")

# 5. Removendo um item
del dados_pais["continente"]

# 6. Iterando sobre o dicionário (útil para criar legendas ou rótulos)
for chave, valor in dados_pais.items():
    print(f"{chave}: {valor}")
```
