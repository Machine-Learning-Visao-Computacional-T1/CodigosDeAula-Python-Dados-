# Introdução ao Pandas e ao conceito de DataFrame
- Pandas é uma biblioteca Python fundamental para análise e manipulação eficiente de dados em diversas áreas.
- DataFrames facilitam a limpeza, transformação e análise exploratória de dados de forma intuitiva.
- Pandas integra-se nativamente com Numpy e ferramentas de visualização para análise completa de dados.

## Definição e estrutura dos DataFrames
- DataFrames organizam dados em linhas e colunas rotuladas, facilitando a visualização e manipulação eficiente.
- Cada coluna pode conter diferentes tipos de dados, permitindo flexibilidade na análise e armazenamento.
- As linhas representam observações.
- Suporte a filtragem, agrupamento e agregação com sintaxe intuitiva e eficiente para análise de dados.
- DataFrames combinam rotulagem e tipos heterogêneos por coluna, facilitando manipulação tabular avançada.

### Importação da biblioteca Pandas

```python
import pandas as pd
```

### Criação de DataFrames

```python
# Criação de um DataFrame manual (através de um Dicionário)
dados = {
    'pais': ['Brasil', 'Argentina', 'Chile'],
    'pib': [1.6, 0.48, 0.31],
    'expectativa_vida': [75.9, 76.7, 80.2]
}
df = pd.DataFrame(dados)

# Importação de arquivos
# df = pd.read_csv('dados.csv')
# df = pd.read_excel('dados.xlsx')
```

```python
df
```

### Exibição de DataFrames

```python
# Exibição e Inspeção Inicial
#print(df.head())      # Primeiras 5 linhas
#print(df.info())      # Estrutura, tipos de dados e memória
print(df.describe())  # Resumo estatístico (média, desvio, etc)
```

### Seleção, Filtros e Indexação
- A manipulação de dados envolve selecionar partes específicas da tabela usando nomes de colunas ou índices

```python
# Seleção por nome de coluna
pibs = df[['pais','pib']]
pibs
```

```python
# loc (por rótulo) e iloc (por posição numérica)
print(df.loc[[0, 2], ['pais', 'pib']])  # Linha 0, coluna 'pais'
#print(df.iloc[1:3, 0:2])  # Primeiras duas linhas e colunas
```

```python
# Seleção por nome de coluna
pibs = df['pib']

# loc (por rótulo) e iloc (por posição numérica)
print(df.loc[0, 'pais'])  # Linha 0, coluna 'pais'
print(df.iloc[0:2, 0:2])  # Primeiras duas linhas e colunas

# Aplicação de Filtros (Boolean Indexing)
# Seleciona apenas países com expectativa de vida acima de 76
filtro = df[df['expectativa_vida'] > 76]
```

### Transformação para Tidy Data
- O conceito de Tidy Data exige que cada variável seja uma coluna e cada observação seja uma linha.
- Frequentemente, dados chegam em formato "largo" e precisam ser convertidos para o formato "longo".

```python
# Exemplo de 'gather' ou 'melt' para tornar os dados 'tidy'
df_tidy = df.melt(id_vars=['pais'], var_name='indicador', value_name='valor')
```

```python
df_tidy
```

# Criação de Colunas e Funções Aplicadas
- Podemos "mutar" o DataFrame criando novas variáveis baseadas em cálculos ou funções complexas.

```python
# Criação simples de coluna
df['pib_per_capita'] = df['pib'] / 200 # Valor hipotético

# Uso de apply(), map() e Lambda
# Transforma nomes de países para maiúsculas
df['pais_maiusculo'] = df['pais'].apply(lambda x: x.upper())
```

```python
df
```

# Estatística, Valores Ausentes e Agrupamento
- O tratamento de dados frequentemente exige lidar com valores designados como NA (dados ausentes) e realizar agregações para resumos.

```python
novalinha = pd.DataFrame({"pais":["Alemanha"], "pib": [5.4]})
df = pd.concat([df, novalinha], ignore_index=True)
```

```python
df
```

```python
# Tratamento de valores ausentes
df.dropna(inplace=True) # Remove linhas com NA
#df.fillna(0, inplace=True) # Preenche com zero

# Operações matemáticas e estatísticas
#media_vida = df['expectativa_vida'].mean()

# Agrupamento (Group By)
# Agrupa por uma categoria e calcula a média das outras colunas
#resumo = df.groupby('pais').mean()
```

```python
resumo = df.groupby('pais').mean()
```

```python
resumo
```

# Gráficos com Matplotlib a partir do Pandas
- A visualização é o passo final para revelar estruturas que estatísticas isoladas podem esconder, como demonstrado pelo Quarteto de Anscombe.
- O Pandas integra-se ao Matplotlib para gerar visualizações rápidas

```python
import matplotlib.pyplot as plt

# Gráfico de barras a partir do DataFrame
df.plot(kind='bar', x='pais', y='pib', color='skyblue')
plt.title("PIB por País")
plt.ylabel("Trilhões de USD")
plt.show()

# Gráfico de dispersão para relacionamentos
df.plot(kind='scatter', x='pib', y='expectativa_vida')
plt.title("Relação PIB vs Expectativa de Vida")
plt.show()
```

# base de Doação de Órgãos, citada nos materiais de Kieran Healy

```python
import pandas as pd
import plotly.express as px
```

```python
# Definindo a URL do arquivo
url = "https://raw.githubusercontent.com/kjhealy/viz-organdata/master/organdonation.csv"

# Lendo o arquivo CSV diretamente da web
df = pd.read_csv(url)
```

```python
# Inspeção Inicial (Olhando para os Dados) Conforme as fontes, é fundamental entender a estrutura antes de visualizar.
#df.head()      # Mostra as 5 primeiras linhas [5]
#df.info()      # Mostra os tipos de colunas e valores ausentes [5]
df.describe()  # Resumo estatístico das variáveis numéricas [5]
```

```python
fig1 = px.scatter(df, x="gdp", y="donors", color="world", 
                 hover_name="country", title="PIB vs Taxa de Doadores")
fig1.show()
```

```python
# Filtrando apenas os dados do Brasil
df_brasil = df[df['country'] == 'Brazil']

fig2 = px.line(df_brasil, x="year", y="donors", 
              title="Evolução da Doação de Órgãos no Brasil")
fig2.show()
```
