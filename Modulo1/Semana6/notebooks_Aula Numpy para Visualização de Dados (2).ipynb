### 1. Criando a Base para os Exemplos

```python
import numpy as np

# Criando um array com 10 elementos (0 a 9)
dados = np.array(range(3,12))
dados
```

### 2. Indexação (Indexing)
- A indexação serve para acessar um único valor específico dentro do array. Lembre-se que em Python a contagem começa sempre em 0.

```python
# Acessando o primeiro elemento
primeiro = dados[0]  # Saída: 3

# Acessando o quarto elemento
quarto = dados[3]    # Saída: 6

# Acessando o último elemento (índice negativo)
ultimo = dados[-1]   # Saída: 11
```

```python
print(primeiro)
```

### 3. Fatiamento (Slicing)
- O fatiamento permite extrair uma subsequência (um pedaço) do array. A sintaxe básica é array[início:fim], onde o índice de fim não é incluído no resultado.

```python
# Pegando do segundo ao quinto elemento (índices 1 a 4)
fatia_meio = dados[1:5]  # Saída: [4-7]

# Pegando os 3 primeiros elementos
primeiros_3 = dados[:3]  # Saída: [3-5]

# Pegando do sexto elemento até o final
final = dados[5:]        # Saída: [8-11]
```

```python
print(final)
```

### 4. Máscaras Booleanas (Masking)
- Esta é uma das técnicas mais poderosas para visualização de dados. Ela permite filtrar o array com base em uma condição lógica, criando uma "máscara" de valores Verdadeiro/Falso.

```python
# Criando uma condição: quais valores são maiores que 50?
condicao = dados > 7
# Se você imprimir 'condicao', verá: [False, False, False, False, False, True, True, True, True, True]

# Aplicando a máscara para filtrar os dados
dados_filtrados = dados[condicao] 
print(dados_filtrados) # Saída: [8-11]

# Exemplo prático: Filtrar apenas valores pares
apenas_pares = dados[dados % 2 == 0] # Filtra múltiplos de 2
```

```python
print(apenas_pares)
```

### 5. Operações Vetorizadas (Aritmética Simples)
- Nas operações vetorizadas, você evita o uso de laços (loops) para processar cada número individualmente. O NumPy aplica a operação a todo o array simultaneamente.

```python
import numpy as np

# Criando um array de base
precos = np.array([10.50, 25.00, 15.75, 40.00])

# Operação Vetorizada: Multiplicação por Escalar (ex: aplicar 10% de desconto)
# O NumPy "propaga" o valor 0.9 por todo o array (lógica similar ao 'recycling' em R) [3]
precos_com_desconto = precos * 0.9
print(f"Preços com desconto: {precos_com_desconto}")

# Operação Vetorizada: Entre Arrays do mesmo tamanho (ex: somar taxas fixas a cada item)
taxas = np.array([1.5, 2.0, 1.2, 3.0])
total = precos + taxas
print(f"Total por item: {total}")
```

### 2. Broadcasting (Difusão)
- O Broadcasting é um recurso do NumPy que permite realizar operações aritméticas entre arrays de formas (shapes) diferentes, desde que eles sigam certas regras de compatibilidade. Informação externa às fontes: essencialmente, o array menor é "transmitido" ou esticado virtualmente para preencher as dimensões do maior.


```python
# Exemplo 1: Somar um valor a uma matriz inteira
matriz_dados = np.array([[5, 7], 
                         [8, 10]])
print(matriz_dados)
# O número 10 (escalar) sofre 'broadcast' para o formato (2, 2)
resultado_escalar = matriz_dados + 10 
print("Matriz + 10:\n", resultado_escalar)

# Exemplo 2: Somar um vetor a cada linha de uma matriz
vetor_ajuste = np.array([11, 13])

# O vetor (11, 13) é transmitido sobre as duas linhas da matriz (2, 2)
ajustado = matriz_dados + vetor_ajuste
print("\nMatriz ajustada por linha:\n", ajustado)
```

# Matplotlib
## Para gerar gráficos com Matplotlib a partir de arrays NumPy, o primeiro passo é configurar os dados e importar as bibliotecas necessárias
- O comando mais fundamental é o plt.plot(x, y), utilizado para criar gráficos de linha simples que mostram a evolução ou funções matemáticas
- Para explorar o relacionamento entre duas variáveis numéricas, o gráfico de dispersão (plt.scatter) é o mais indicado
- Quando o objetivo é comparar categorias ou quantidades, utilizamos o gráfico de barras com o comando plt.bar
- Já para visualizar a distribuição de probabilidade de um conjunto de dados, o comando plt.hist gera histogramas de forma eficiente
- Abaixo estão exemplos práticos de como implementar essas visualizações

### 1. Gráfico de Linha (Evolução)

```python
import matplotlib.pyplot as plt
import numpy as np

# Criando arrays NumPy
x = np.linspace(0, 10, 100)  # Gera 100 pontos entre 0 e 10 [1]
y = np.sin(x)               # Calcula o seno de cada ponto do array [1]

plt.plot(x, y, label='Seno')
plt.title("Gráfico de Linha Simples") # Adicionando título [1]
plt.xlabel("Tempo")                  # Rótulo do eixo X [1]
plt.ylabel("Amplitude")              # Rótulo do eixo Y [1]
plt.legend()
plt.show()
```

### 2. Gráfico de Dispersão (Relações)

```python
# Gerando 50 pontos aleatórios com NumPy para os eixos X e Y [3]
x = np.random.rand(50)
y = np.random.rand(50)

# Criando o gráfico de dispersão com pontos roxos [3]
plt.scatter(x, y, color='purple')
plt.title("Gráfico de Dispersão a partir de NumPy")
plt.show()
```

### 3. Gráfico de Barras (Quantidades e Proporções)

```python
# Definindo rótulos e valores quantitativos em arrays [4]
categorias = np.array(['Setor A', 'Setor B', 'Setor C'])
valores = np.array([5, 6, 7])

plt.bar(categorias, valores, color='seagreen') # Criação das barras [4]
plt.title("Comparação entre Setores")
plt.ylabel("Valores")
plt.show()
```

### 4. Histograma (Distribuição)

```python
# Criando uma amostra de 1000 números seguindo uma distribuição normal [4]
amostra = np.random.randn(1000)

plt.hist(amostra, bins=20, edgecolor='black') # Agrupando em 20 bins [4]
plt.title("Análise de Distribuição (Histograma)")
plt.show()
```
