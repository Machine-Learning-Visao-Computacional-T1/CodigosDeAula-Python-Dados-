```python
ativo = True
if ativo:
    print("Ativo")
else:
    print("Inativo")
```

```python
if ativo == False:
    print("Ativo")
else:
    print("Inativo")
```

```python
inativo = False
if (ativo | inativo):
    print("Ativo")
else:
    print("Inativo")


inativo = False
if (ativo & inativo):
    print("Ativo")
else:
    print("Inativo")
```

```python
a = 10
b = 20
if a > 11 or b > 19:
    print("or sim")
else:
    print("or não")

if a > 11 | b > 19:
    print("| sim")
else:
    print("| não ")

if a < 11 and b > 19:
    print("and sim")
else:
    print("and não")

if a < 11 & b > 19:
    print("& sim")
else:
    print("& não ")

if b > 19 & a < 9:
    print("& sim")
else:
    print("& não ")
```

```python
def calcular_imc(peso, altura):
    return peso / (altura ** 2)

meu_imc = calcular_imc(80, 1.80)

print(f"Meu IMC é: {meu_imc:.2f}")


def definicao_imc(meu_imc):
  if meu_imc < 18.5:
    print("Abaixo do peso")
  elif meu_imc < 24.9:
    print("Peso normal")
  elif meu_imc < 29.9:
    print("Sobrepeso")
  else:
    print("Obesidade")

definicao_imc(meu_imc)
```

```python
def somar(a, b):
    resultado = a + b
    return resultado # A função retorna o valor de 'resultado'

# Chamando a função e armazenando o valor retornado em uma variável
soma_total = somar(5, 3)

# Usando o valor retornado
print(f"A soma é: {soma_total}")

# Podemos usar o valor retornado em outras operações
multiplicacao = soma_total * 2
print(f"O dobro da soma é: {multiplicacao}")
```
