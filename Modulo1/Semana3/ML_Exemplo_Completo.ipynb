```python
# 1. IMPORTANDO AS FERRAMENTAS
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
print("--- 1. CARREGANDO E MANIPULANDO DADOS (A Lógica em Ação) ---")
dataset = load_wine()
X_bruto = dataset.data   # As características dos vinhos (Álcool, Cor, etc)
y_bruto = dataset.target # A classe do vinho (0, 1 ou 2)

print("\n1. Primeiras 5 linhas do X_bruto (Características):")
# O [:5] diz: "pegue do início até o índice 4"
print(X_bruto[:5])

# Printando os 5 primeiros alvos (y)
print("\n2. Primeiros 5 labels do y_bruto (Gabarito Real):")
print(y_bruto[:5])
```

```python
print("--- EXIBINDO O 'HEAD' COMPLETO (10 PRIMEIROS EXEMPLOS) ---")

# 1. Pegamos todos os nomes das colunas
nomes_colunas = dataset.feature_names

# 2. Criamos o cabeçalho juntando os nomes com uma barra vertical
# Usamos o .upper() para dar destaque, como vimos em Strings!
cabecalho = " | ".join([nome.upper() for nome in nomes_colunas]) + " | [CLASSE]"
print(cabecalho)
print("-" * len(cabecalho)) # Cria uma linha do tamanho exato do cabeçalho

# 3. Loop para os 10 primeiros exemplos
for i in range(10):
    # Usamos List Comprehension para arredondar cada número para 2 casas decimais
    # Isso evita que a tabela fique desalinhada com números gigantes
    linha_formatada = [f"{valor:.2d}" if isinstance(valor, int) else f"{valor:.2f}" for valor in X_bruto[i]]

    # Juntamos os valores formatados e printamos com a classe no final
    print(" | ".join(linha_formatada) + f" | {y_bruto[i]}")
```

```python
print(dataset['DESCR'])
```

```python
# a) Usando List Comprehension e Condicionais para binarizar o alvo:
# Se for o vinho tipo 0, será classe 1 (Premium). O resto será 0 (Padrão).
y_processado = np.array([1 if classe == 0 else 0 for classe in y_bruto])
```

```python
# b) Engenharia de Atributos com Laço FOR e IF/ELSE!
# Vamos criar uma nova coluna: "Teor Alcoólico Alto" (1 para Sim, 0 para Não)
# O nível de Álcool é a coluna de índice 0 no X_bruto.
X_novo = []
for linha in X_bruto:
    teor_alcool = linha[0]

    if teor_alcool >= 13.5:
        flag_alcool_alto = 1
    else:
        flag_alcool_alto = 0

    # Transformamos a linha em lista, adicionamos a nova coluna e guardamos
    linha_lista = list(linha)
    linha_lista.append(flag_alcool_alto)
    X_novo.append(linha_lista)
```

```python
# Convertendo de volta para Array do NumPy (Acelerador matemático)
X_processado = np.array(X_novo)

print(f"Dados originais: {X_bruto.shape[1]} colunas.")
print(f"Dados processados: {X_processado.shape[1]} colunas (Nova coluna foi adicionada!).\n")
```

```python
print(X_processado[:5])
```

```python
print(y_processado[:5])
```

```python
print("--- O DATASET PROCESSADO ---")

# 1. Pegamos os nomes originais do dataset
nomes_colunas = list(dataset.feature_names)

# 2. ADICIONAMOS A NOSSA NOVA COLUNA NO FINAL DA LISTA DE NOMES!
nomes_colunas.append("teor_alcool_alto")

# 3. Criamos o cabeçalho atualizado
cabecalho = " | ".join([nome.upper() for nome in nomes_colunas]) + " | [CLASSE_BINARIA]"
print(cabecalho)
print("-" * len(cabecalho)) # Cria a linha divisória do tamanho exato

# 4. Loop para os 10 primeiros exemplos usando os dados NOVOS
for i in range(10):
    # Usamos List Comprehension para formatar os valores de X_processado
    # Como a nossa nova coluna é 0 ou 1, podemos deixá-la visualmente mais limpa
    linha_formatada = [f"{int(valor)}" if j == 13 else f"{valor:.2f}" for j, valor in enumerate(X_processado[i])]

    # Juntamos tudo e printamos com o novo gabarito (y_processado) no final
    print(" | ".join(linha_formatada) + f" | {y_processado[i]}")
```

```python
print("--- 2. SEPARANDO TREINO E TESTE ---")
# Separamos 30% dos dados para testar a IA depois de treinada (para ela não "decorar" as respostas)
X_treino, X_teste, y_treino, y_teste = train_test_split(X_processado, y_processado, test_size=0.90, random_state=42)
print(f"Vinhos para Treinar a IA: {len(X_treino)}")
print(f"Vinhos para Testar a IA: {len(X_teste)}\n")


# Na etapa 2, mude o test_size de 0.30 para 0.90:
# X_treino, X_teste, y_treino, y_teste = train_test_split(X_processado, y_processado, test_size=0.90, random_state=42)
```

```python
print("--- 3. TREINANDO O MODELO (A Mágica) ---")
# Escolhemos o Algoritmo "Floresta Aleatória"
modelo_ia = RandomForestClassifier(random_state=42)
modelo_ia.fit(X_treino, y_treino) # O comando FIT é a IA aprendendo!
print("Inteligência Artificial treinada com sucesso!\n")


# Na etapa 3, treine a IA cortando a última coluna (nossa engenharia de atributos)
# O comando [:, :-1] significa "pegue todas as linhas, mas remova a última coluna"
#modelo_ia.fit(X_treino[:, :-1], y_treino)

# Na etapa 4, corte a última coluna na hora da previsão também:
#previsoes = modelo_ia.predict(X_teste[:, :-1])
```

```python
print("--- 4. FAZENDO PREVISÕES (Predict) ---")
# Pedimos para a IA prever as respostas baseada apenas nas características do X_teste
previsoes = modelo_ia.predict(X_teste)
```

```python
print("--- 5. AVALIANDO O DESEMPENHO (Métricas) ---")
# Comparamos as previsões da IA com o gabarito real (y_teste)
acuracia = accuracy_score(y_teste, previsoes)
precisao = precision_score(y_teste, previsoes)
recall = recall_score(y_teste, previsoes)
f1 = f1_score(y_teste, previsoes)
```

```python
print(f"🎯 Acurácia: {acuracia * 100:.2f}%")
print(f"🎯 Precisão: {precisao * 100:.2f}%")
print(f"🎯 Recall:   {recall * 100:.2f}%")
print(f"🎯 F1-Score: {f1 * 100:.2f}%\n")
```

```python
print("--- 6. MATRIZ DE CONFUSÃO ---")
matriz = confusion_matrix(y_teste, previsoes)
print(matriz)
print("\nLegenda da Matriz:")
print("[ Verdadeiros Negativos (VN)  |  Falsos Positivos (FP) ]")
print("[ Falsos Negativos (FN)       |  Verdadeiros Positivos (VP) ]")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("--- 5. RELATÓRIO DE CLASSIFICAÇÃO  ---")
# O target_names substitui os números 0 e 1 por textos para ficar mais legível
relatorio = classification_report(y_teste, previsoes, target_names=['Vinho Padrão (0)', 'Vinho Premium (1)'])
print(relatorio)

print("--- 6. MATRIZ DE CONFUSÃO ---")
matriz = confusion_matrix(y_teste, previsoes)
print(matriz)
```

```python
from sklearn.model_selection import cross_val_score

print("--- VALIDAÇÃO CRUZADA ---")

# rodar o modelo 5 vezes, trocando quem é treino e quem é teste!

notas_validacao = cross_val_score(modelo_ia, X_processado, y_processado, cv=5)

print(f"Notas de acurácia em cada uma das 5 rodadas: {notas_validacao}")
print(f"Média Final de Acertos: {notas_validacao.mean() * 100:.2f}%")
print(f"Margem de Erro (Desvio Padrão): ±{notas_validacao.std() * 100:.2f}%\n")

print("--- (Feature Importance) ---")
# Vamos ver quais colunas ela achou mais importantes para tomar a decisão
importancias = modelo_ia.feature_importances_

# Pegamos os nomes originais e adicionamos a nossa coluna
nomes_completos = list(dataset.feature_names)
nomes_completos.append("teor_alcool_alto")

# Mostrando o Top 3 características mais usadas
print("Top 3 Regras que a IA usou para separar os vinhos:")
dicionario_importancias = dict(zip(nomes_completos, importancias))
top_3 = sorted(dicionario_importancias.items(), key=lambda x: x[1], reverse=True)[:3]

for nome, valor in top_3:
    print(f"- {nome.upper()}: {valor * 100:.1f}% de importância")
```
