```python
from ultralytics import YOLO
import urllib.request
from google.colab.patches import cv2_imshow

# 1. Baixa uma foto de teste genérica da internet
url = "https://ultralytics.com/images/zidane.jpg"
urllib.request.urlretrieve(url, "teste.jpg")

# 2. Carrega o modelo de fábrica (já treinado para achar pessoas)
model = YOLO("yolo11n.pt")

# 3. Faz a detecção
resultados = model.predict("teste.jpg")

# 4. Mostra a imagem com os quadrados na tela do Colab
res_plot = resultados[0].plot()
cv2_imshow(res_plot)

print("✅ Código executado com sucesso! teste")
```
