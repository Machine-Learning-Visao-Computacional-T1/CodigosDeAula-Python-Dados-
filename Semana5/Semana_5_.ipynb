## Padrão de definição em python para engenharia de dados:

```python
TAXA = 0.15
```

```python
def calcular_imposto(valor: float, taxa: float = 0.15) -> float:
    """
    Calcula o imposto sobre um valor com uma taxa padrão de 15%.
    Retorna o valor final com o imposto aplicado.
    """
    imposto = valor * taxa
    valor_final = valor + imposto
    return valor_final # O 'return' É preferivel retornar em variável

# Usando a função:
preco_notebook = calcular_imposto(3500.00,TAXA)
print(f"Preço final: R$ {preco_notebook}")
```

## Função Lambda:

```python
# Função Clássica
def quadrado(x):
    return x ** 2

# Mesma função, versão Lambda
quadrado_lambda = lambda x: x ** 2

print(quadrado_lambda(5)) # Saída: 25

# APLICAÇÃO REAL: Usando Lambda com a função map() para aplicar a regra em uma lista inteira
precos_sujos = [10.50, 20.00, 35.99]
precos_com_desconto = list(map(lambda p: p * 0.90, precos_sujos))

print(precos_com_desconto) # Saída: [9.45, 18.0, 32.391]
```

## Expressões Regulares(REGEX)

```python
import re

texto_chat = """
Atendente: O notebook gamer custa R$ 4.500,00 na loja física.
Cliente: Mas no site vi por R$3999,90!
Atendente: Consigo fazer por R$ 4.000,00 se pagar à vista.
"""

# 1. Encontrando os padrões de dinheiro no texto
padrao_dinheiro = r'R\$\s*[\d\.,]+'
precos_brutos = re.findall(padrao_dinheiro, texto_chat)

print("Preços encontrados no chat:", precos_brutos)
# Saída: ['R$ 4.500,00', 'R$3999,90', 'R$ 4.000,00']

# 2. Transformando em Float (Opcional, mas nível Sênior!)
precos_matematicos = []
for p in precos_brutos:
    # Remove R$, espaços e pontos de milhar, e troca a vírgula por ponto decimal
    p_limpo = re.sub(r'[R\$\s\.]', '', p).replace(',', '.')
    precos_matematicos.append(float(p_limpo))

print("Preços prontos para o Numpy:", precos_matematicos)
```

## Vetorização e Broadcasting

```python
import numpy as np

precos = np.array([100.0, 250.0, 300.0])

# Sem Vetorização (Lento - Estilo Python Clássico)
precos_com_juros = []
for p in precos:
    precos_com_juros.append(p * 1.10)
print(precos_com_juros)

# COM Vetorização e Broadcasting (Rápido - Estilo Numpy Sênior)
# Multiplicamos o array inteiro por um número escalar. O Numpy "estica" o 1.10 para todos!
precos_finais = precos * 1.10

print(precos_finais) # Saída: [110. 275. 330.]
```

```python
import numpy as np

# Broadcasting: Multiplicando preços de produtos em 2 lojas diferentes por taxas de impostos
# 2 Lojas (Linhas), 3 Produtos (Colunas)
precos_lojas = np.array([
    [100, 200, 300],
    [150, 250, 350]
])

# Taxas de imposto para os 3 produtos (Vetor 1D)
taxas = np.array([1.10, 1.15, 1.05])

# O Numpy "estica" o vetor de taxas para as duas lojas automaticamente!
precos_finais = precos_lojas * taxas

print("Preços com imposto aplicado em todas as lojas simultaneamente:\n", precos_finais)
```

## Matriz e Reshape

```python
import numpy as np

# Temos um vetor de 6 números (Array 1D)
dados = np.array([10, 20, 30, 40, 50, 60])
print("Dimensão original:", dados.shape) # (6,)

# O Algoritmo de ML exige 2D (Linhas e Colunas).
# O método .reshape() remodela a matriz sem alterar os dados.
# O '-1' pede para o Numpy calcular automaticamente as linhas, criando 1 coluna
dados_para_ml = dados.reshape(-1, 1)

print("Dimensão para ML:\n", dados_para_ml)
# Saída: Matriz com 6 linhas e 1 coluna (2D)
print("Dimensão para ML reshape:", dados_para_ml.shape)
```

```python
import numpy as np

# Simulando um Dataset: [Idade, Salário, Anos_Estudo, Comprou?]
dataset = np.array([
    [25, 5000, 4, 1],
    [30, 8000, 6, 0],
    [45, 12000, 8, 1],
    [22, 2500, 2, 0]
])

print("--- SEPARANDO OS DADOS ---")
# X: Todas as linhas (:), Todas as colunas exceto a última (:-1)
X = dataset[:, :-1]

# y: Todas as linhas (:), Apenas a última coluna (-1)
y = dataset[:, -1]

# =====================================================================
# A PROVA MATEMÁTICA (Analisando as Dimensões)
# O atributo .ndim mostra a dimensão real do array.
# O atributo .shape mostra o formato exato (Linhas, Colunas).
# =====================================================================

print("\n ANALISANDO O 'X' (Características):")
print(X)
print(f"-> Dimensão de X: {X.ndim}D")
print(f"-> Formato de X : {X.shape} (4 Linhas, 3 Colunas)")
print("Aprovado para o Scikit-Learn! É uma Matriz bidimensional.")

print("\n ANALISANDO O 'y' (Alvo/Target):")
print(y)
print(f"-> Dimensão de y: {y.ndim}D")
print(f"-> Dimensao de y após reshape: {y.reshape(-1, 1).ndim}")
print(f"-> Formato de y : {y.shape} (4 Elementos, sem largura de coluna)")
print(f"-> Formato de y após reshape: {y.reshape(-1, 1).shape}")
print("Aprovado para o Scikit-Learn! É um Vetor unidimensional.")
```

### Comparando Leitura com Pandas e Numpy

```python
import numpy as np
import pandas as pd
import io

# =====================================================================
# OS NOSSOS DOIS ARQUIVOS SIMULADOS
# =====================================================================
csv_sem_cabecalho = """25,5000.0,1
30,8000.0,0
45,12000.0,1"""

csv_com_cabecalho = """Idade,Salario,Comprou
25,5000.0,1
30,8000.0,0
45,12000.0,1"""

print("==================================================")
print(" 🥊 ROUND 1: LENDO DADOS SEM CABEÇALHO (Pura Matemática)")
print("==================================================")

# NUMPY: Lê perfeitamente, pois só tem números!
matriz_numpy_sem = np.loadtxt(io.StringIO(csv_sem_cabecalho), delimiter=',')
print("\n🟢 NUMPY (Sem Cabeçalho): Sucesso absoluto!")
print(matriz_numpy_sem)
print(type(matriz_numpy_sem))

# PANDAS: Vai errar! Como não avisamos, ele "rouba" a primeira linha de dados para virar o título da coluna!
df_pandas_errado = pd.read_csv(io.StringIO(csv_sem_cabecalho))
print("\n🔴 PANDAS (Sem Cabeçalho): Errou! Transformou o cliente de 25 anos em título de coluna:")
display(df_pandas_errado)

# CORREÇÃO PANDAS: Precisamos avisar 'header=None'
df_pandas_corrigido = pd.read_csv(io.StringIO(csv_sem_cabecalho), header=None)
print("\n🟢 PANDAS (Sem Cabeçalho - Corrigido com header=None):")
display(df_pandas_corrigido)


print("\n\n==================================================")
print(" 🥊 ROUND 2: LENDO DADOS COM CABEÇALHO (O Mundo Real)")
print("==================================================")

# NUMPY: Se rodarmos direto, ele quebra! (ValueError: could not convert string to float: 'Idade')
# Precisamos OBRIGATORIAMENTE pular a primeira linha (skiprows=1)
matriz_numpy_com = np.loadtxt(io.StringIO(csv_com_cabecalho), delimiter=',', skiprows=1)
print("\n🟡 NUMPY (Com Cabeçalho): Só funciona se usarmos 'skiprows=1' para ignorar o texto:")
print(matriz_numpy_com)

# PANDAS: Aqui ele brilha! É o comportamento padrão dele.
df_pandas_com = pd.read_csv(io.StringIO(csv_com_cabecalho))
print("\n🟢 PANDAS (Com Cabeçalho): Feito para isso! Reconheceu os nomes das colunas automaticamente:")
display(df_pandas_com)
```

### Filtros e Seleções

```python
# =====================================================================
# 2. FILTROS E SELEÇÕES (.loc, .iloc e Máscaras)
# =====================================================================
print("\n--- 1. Filtro Simples (Valor > 5000) ---")
vendas_altas = df_vendas[df_vendas['Valor'] > 5000]
display(vendas_altas)

print("\n--- 2. Múltiplas Condições (Idade > 30 E Estado == MG) ---")
clientes_alvo = df_vendas[(df_vendas['Idade'] > 30) & (df_vendas['Estado'] == 'MG')]
display(clientes_alvo)

print("\n--- 3. Filtro com .loc (Estado == MG, mostrando só Cliente e Valor) ---")
resumo_mg = df_vendas.loc[df_vendas['Estado'] == 'MG', ['Cliente', 'Valor']]
display(resumo_mg)

print("\n--- 4. Fatiamento com .iloc (Primeiras 3 linhas, Primeiras 3 colunas) ---")
amostra_dados = df_vendas.iloc[0:3, 0:3]
display(amostra_dados)
```

### Datetime - Datas

```python
# =====================================================================
# 3. TRABALHANDO COM DATAS (Datetime)
# =====================================================================
# Convertendo texto para data matemática
df_vendas['Data_Venda'] = pd.to_datetime(df_vendas['Data_Venda'], format='%d/%m/%Y')
df_vendas['Mes'] = df_vendas['Data_Venda'].dt.month # Extraindo apenas o mês
df_vendas['Ano'] = df_vendas['Data_Venda'].dt.year # Extraindo apenas o ano

print(df_vendas)
print(type(df_vendas))
print(type(df_vendas['Data_Venda']))
df_vendas.info()
df_vendas.head()
```

## Operacoes Relacionais

```python
# =====================================================================
# 4. GROUPBY (Agrupamento e Resumo)
# =====================================================================
print("\n--- 5. Faturamento Total por Estado (GroupBy) ---")
faturamento_estado = df_vendas.groupby('Estado')['Valor'].sum().reset_index()
display(faturamento_estado)

print('\n Faturamento total por mes/ano')
faturamento_mes_ano = df_vendas.groupby(['Mes', 'Ano'])['Valor'].sum().reset_index()
display(faturamento_mes_ano)
```

```python
# =====================================================================
# 5. MERGE (Cruzamento Relacional / PROCV)
# =====================================================================
print("\n--- 6. Cruzando Vendas com Status dos Clientes (Merge) ---")
df_final = pd.merge(df_vendas, df_clientes, on='ID_Cliente', how='left')
display(df_final)
```

## Pivot Table

```python
# =====================================================================
# 6. PIVOT TABLE (Tabela Dinâmica / Visão Matricial)
# =====================================================================
print("\n--- 7. Tabela Dinâmica: Clientes (Linhas) vs Meses (Colunas) ---")
matriz_vendas = df_final.pivot_table(
    values='Valor',
    index='Cliente',
    columns='Mes',
    aggfunc='sum',
    fill_value=0
)
display(matriz_vendas)
```
